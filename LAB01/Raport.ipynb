## 1. Qëllimi i Punës
* Njohja me gjuhën e shënimeve **Markdown** për krijimin e raporteve shkencore dhe teknike të strukturuara qartë dhe të lexueshme direkt në platforma si GitHub.
* Kuptimi i parimit të punës, strukturës së brendshme dhe mënyrës së përdorimit të **Breadboard-it** (Pllakës së Provave) si mjet bazë për ndërtimin e qarqeve elektronike të përkohshme pa saldim.

## 2. Pjesa e Parë: Gjuha Markdown
**Markdown** është një gjuhë shënimi e lehtë me një sintaksë të thjeshtë teksti, e cila lejon konvertimin e dokumentit në formate të ndryshme. Në këtë lëndë, ajo përdoret për të dokumentuar eksperimentet dhe detyrat laboratorike.

### Elementët Kryesorë të Përdorur:
* **Titujt (`#`, `##`, `###`):** Për të krijuar një strukturë hierarkike të qartë të raportit.
* **Formatimi i tekstit:** Përdorimi i *pjerrtësisë* (`*italic*`), **trashësisë** (`**bold**`), dhe blloqeve të kodit (`code blocks`).
* **Listat dhe Tabelat:** Për të organizuar të dhënat dhe hapat e eksperimentit në mënyrë të lexueshme.
* **Lidhjet dhe Imazhet:** Për të integruar skemat e qarqeve dhe grafikat e krijuara gjatë matjeve.

## 3. Pjesa e Dytë: Breadboard (Pllaka e Provave)
**Breadboard** është një pajisje pa saldim (prototyping board) që përdoret për të ndërtuar dhe testuar qarqe elektronike. Ajo lejon modifikimin e shpejtë të lidhjeve gjatë fazës së testimit.

### Struktura e Brendshme e Breadboard-it:
1. **Linjat e Furnizimit (Power Rails - Pjesët Anësore):**
   * Zakonisht të shënuara me vija të kuqe (`+`) për tensionin dhe blu/të zeza (`-`) për tokëzimin (GND).
   * Këto vrima janë të lidhura horizontalisht përgjatë gjithë gjatësisë së pllakës për të shpërndarë energjinë në qark.
2. **Blloqet Qendrore (Terminal Strips - Pjesa e Mesme):**
   * Janë të ndara në blloqe vertikale. Vrimat brenda çdo grupi prej 5 vrimash në një rresht vertikal janë të lidhura elektrikisht midis tyre.
3. **Kanalet e Izolimit (Center Notch):**
   * Kanali plastik në mes ndan pllakën në dy gjysma, duke lejuar vendosjen e qarqeve të integruara (IC) pa shkaktuar lidhje të shkurtra mes pin-eve të tyre.

## 4. Përfundime

* Breadboard-i ofron një mjedis fleksibël dhe të sigurt për testimin e prototipeve të qarqeve elektrike para realizimit të tyre përfundimtar.
