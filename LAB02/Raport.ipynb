# 1. Qëllimi i Eksperimentit

Ne kete eksperiment analizohet sjellja e qarqeve rezistive te lidhura ne seri dhe ne paralel ne bordin laboratorik (breadboard). Verifikohet eksperimentalisht parimi i ndarjes se tensionit (Voltage Divider) dhe ligjet e Kirchhoff-it ne regjim te rrymes se vazhduar (DC). Percaktohet tensioni ne dalje $V_{out}$ ne dy raste: kur qarku eshte i hapur (pa ngarkese) dhe kur ne dalje lidhet nje rezistence ngarkese $R_L$. Gjithashtu, testohet marreveshja teorike midis vlerave te llogaritura me formulat e rezistences ekuivalente dhe vlerave matura me multimetrin digjital.


# 2. Baza Teorike

## 2.1 Modeli i Qarkut

Qarku përbëhet nga një burim tensioni hyrës $V_{in}$ dhe dy rezistenca të lidhura në seri ($R_1$ dhe $R_2$). Tensioni dalës $V_{out}$ merret në skajet e rezistencës $R_2$.

- **Sinjali hyrës:** $V_{in}$ (tension i vazhduar DC)
- **Sinjali dalës:** $V_{out}$ (merret në nyjën midis $R_1$ dhe $R_2$ kundrejt masës GND)
- **Vlerat e komponentëve:** Rezistencat $R_1$, $R_2$ dhe rezistenca e ngarkesës $R_L$
- **Supozimet:** Komponentë linearë idealë, regjim stacionar DC, rezistencë e brendshme e burimit e papërfillshme.


## 2.2 Ekuacionet Themelore

Për pjesëtuesin e thjeshtë të tensionit pa ngarkesë (Eksperimenti 1):

$$
\frac{V_{out}}{V_{in}} = \frac{R_2}{R_1 + R_2}
$$

Kur qarkut i shtohet një rezistencë ngarkese $R_L$ në paralel me $R_2$ (Eksperimentet 2 dhe 3), rezistenca ekuivalente e degës së poshtme bëhet:

$$
R_2 \parallel R_L = \frac{R_2 \cdot R_L}{R_2 + R_L}
$$

Dhe tensioni dalës llogaritet me formulën:

$$
V_{out} = \frac{R_2 \parallel R_L}{R_1 + (R_2 \parallel R_L)} \cdot V_{in}
$$

# 4. Të Dhënat e Matura

```python
import pandas as pd
import numpy as np

# Të dhënat e regjistruara në laborator (11.11.2026)
data = pd.DataFrame({
    "Eksperimenti": [
        "Eksp 1 (Pa ngarkesë, kΩ)", 
        "Eksp 2 (Pa ngarkesë, MΩ)", 
        "Eksp 3 - Rasti A (RL = 3.30 MΩ)", 
        "Eksp 3 - Rasti B (RL = 2.34 MΩ)"
    ],
    "R1 (kΩ)": [7.55, 3270.0, 7.55, 7.55],
    "R2 (kΩ)": [7.78, 3300.0, 7.78, 7.78],
    "RL (kΩ)": [np.inf, np.inf, 3300.0, 2340.0],
    "Vin (V)": [15.35, 15.77, 15.35, 15.35],
    "Vout_exp (V)": [7.50, 6.71, 7.76, 7.71]
})

data
```

```python
# Llogaritja e vlerave teorike dhe gabimit relativ
R2_eq = np.where(np.isinf(data["RL (kΩ)"]), data["R2 (kΩ)"], (data["R2 (kΩ)"] * data["RL (kΩ)"]) / (data["R2 (kΩ)"] + data["RL (kΩ)"]))
data["Vout_teori (V)"] = (R2_eq / (data["R1 (kΩ)"] + R2_eq)) * data["Vin (V)"]
data["Gabimi Relativ (%)"] = (np.abs(data["Vout_exp (V)"] - data["Vout_teori (V)"]) / data["Vout_teori (V)"]) * 100

data.round(3)
```

# 5. Analiza e të Dhënave

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 4.5))
x = np.arange(len(data["Eksperimenti"]))
width = 0.35

plt.bar(x - width/2, data["Vout_teori (V)"], width, label="Vout Teorik (V)", color="#4C72B0")
plt.bar(x + width/2, data["Vout_exp (V)"], width, label="Vout Eksperimental (V)", color="#DD8452")

plt.ylabel("Tensioni Vout (V)")
plt.title("Krahasimi i Tensionit Dalës: Teori vs. Eksperiment")
plt.xticks(x, ["Eksp 1", "Eksp 2", "Eksp 3 (A)", "Eksp 3 (B)"])
plt.legend()
plt.grid(axis="y", linestyle="--", alpha=0.7)
plt.tight_layout()
plt.show()
```

Nga grafiku dhe tabela vërejmë:

- **Eksperimenti 1:** Me rezistenca të rendit $\text{k}\Omega$, vlera eksperimentale ($7.50\text{ V}$) përputhet plotësisht me vlerën teorike.  
- **Eksperimenti 2:** Me rezistenca shumë të mëdha ($\sim 3.3\text{ M}\Omega$), tensioni i matur ($6.71\text{ V}$) ndikohet nga rezistenca e brendshme e multimetrit ($R_{multimetër} \approx 10\text{ M}\Omega$).  
- **Eksperimenti 3:** Kur rezistenca e ngarkesës $R_L$ lidhet në paralel me $R_2$, ndikimi në tensionin dalës mbetet brenda parashikimeve.  


# 6. Analiza e Gabimeve

Gabimi relativ u llogarit me formulën:

$$
\text{Gabimi} = \frac{|V_{exp} - V_{teori}|}{V_{teori}} \times 100\%
$$

Burimet kryesore të gabimeve:

- **Rezistenca e brendshme e multimetrit:** Në Eksperimentin 2 ($R_1, R_2 \approx 3.3\text{ M}\Omega$), impedanca hyrëse e multimetrit lidhet në paralel me $R_2$.  
- **Tolerancat e komponentëve:** Rezistencat e përdorura kanë tolerancë standarde $\pm 5\%$.  
- **Pasiguritë e matjes dhe kontaktet:** Rezistenca e telave lidhës dhe kontakteve në pllakën eksperimentale ACT 10.
