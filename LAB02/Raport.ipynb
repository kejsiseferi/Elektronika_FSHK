# 1. Qellimi i Eksperimentit

Ne kete eksperiment analizohet sjellja e qarqeve rezistive te lidhura ne seri dhe ne paralel ne bordin laboratorik (breadboard). Verifikohet eksperimentalisht parimi i ndarjes se tensionit (Voltage Divider) dhe ligjet e Kirchhoff-it ne regjim te rrymes se vazhduar (DC). Percaktohet tensioni ne dalje $V_{out}$ ne dy raste: kur qarku eshte i hapur (pa ngarkese) dhe kur ne dalje lidhet nje rezistence ngarkese $R_L$. Gjithashtu, testohet marreveshja teorike midis vlerave te llogaritura me formulat e rezistences ekuivalente dhe vlerave te matura me multimetrin digjital.


# 2. Baza Teorike

## 2.1 Modeli i Qarkut

Qarku baze perbehet nga dy rezistenca $R_1$ dhe $R_2$ te lidhura ne seri me burimin e tensionit te vazhduar $V_{in}$ dhe token ($Gnd$). Tensioni i daljes $V_{out}$ merret ne nyjen midis dy rezistencave.

- **Sinjali hyres:** Tensioni i vazhduar $V_{in}$ nga burimi DC i trainerit laboratorik.
- **Sinjali dales:** Tensioni $V_{out}$ i matur ne skajet e rezistences $R_2$ (ose ekuivalentit $R_2 \parallel R_L$).
- **Vlerat e komponenteve:** Rezistenca fikse ne rendin e $\text{k}\Omega$ dhe $\text{M}\Omega$ (p.sh. $R_1 = 7.55\text{ k}\Omega$, $R_2 = 7.78\text{ k}\Omega$, $R_L = 3.30\text{ k}\Omega$).
- **Supozimet:** Komponentet konsiderohen rezistenca ideale lineare, percjellesit kane rezistence te paperfillshme dhe burimi DC jep tension konstant.


## 2.2 Ekuacionet Themelore

Per ndaresin e tensionit pa ngarkese (qark i hapur ne dalje), rryma e njejte kalon neper $R_1$ dhe $R_2$, keshtu qe tensioni i daljes jepet nga:

$$
V_{out} = V_{in} \cdot \frac{R_2}{R_1 + R_2}
$$

Kur ne dalje lidhet nje rezistence ngarkese $R_L$ paralelisht me $R_2$, rezistenca ekuivalente e deges se poshtme behet:

$$
R_{ek} = R_2 \parallel R_L = \frac{R_2 \cdot R_L}{R_2 + R_L}
$$

Ne kete rast, tensioni i daljes me ngarkese llogaritet me formulen:

$$
V_{out} = V_{in} \cdot \frac{R_2 \parallel R_L}{R_1 + (R_2 \parallel R_L)}
$$

# 4. Te Dhenat e Matura

```python
import pandas as pd
import numpy as np

# Te dhenat eksperimentale nga fletorja e laboratorit (Eksperimenti 1, 2 dhe 3)
data = pd.DataFrame({
    "Eksperimenti": [
        "Eksp 1: Ndarsi i tensionit (kΩ)",
        "Eksp 2: Qarku i hapur (MΩ)",
        "Eksp 3a: Me ngarkese RL (3.30 kΩ)",
        "Eksp 3b: Me ngarkese RL (2.34 kΩ)"
    ],
    "R1 (kΩ)": [7.55, 3270.0, 7.55, 7.55],
    "R2 (kΩ)": [7.78, 3300.0, 7.78, 7.78],
    "RL (kΩ)": [np.nan, np.nan, 3.30, 2.34],
    "Vin (V)": [15.35, 15.77, 15.35, 15.35],
    "Vout_exp (V)": [7.50, 6.71, 7.76, 7.71]
})

data
```

```python
# Llogaritja e rezistences ekuivalente R2 || RL dhe tensionit teorik Vout_teori
def llogarit_vout_teori(row):
    if pd.isna(row["RL (kΩ)"]):
        r_ek = row["R2 (kΩ)"]
    else:
        r_ek = (row["R2 (kΩ)"] * row["RL (kΩ)"]) / (row["R2 (kΩ)"] + row["RL (kΩ)"])
    return row["Vin (V)"] * (r_ek / (row["R1 (kΩ)"] + r_ek))

data["Vout_teori (V)"] = data.apply(llogarit_vout_teori, axis=1).round(2)
data["Gabimi Relativ (%)"] = (np.abs(data["Vout_exp (V)"] - data["Vout_teori (V)"]) / data["Vout_teori (V)"] * 100).round(2)

data
```

# 5. Analiza e te Dhenave

```python
import matplotlib.pyplot as plt

x = np.arange(len(data["Eksperimenti"]))
width = 0.35

plt.figure(figsize=(8, 4.5))
plt.bar(x - width/2, data["Vout_teori (V)"], width, label="Vout Teorik (V)", color="#4C72B0")
plt.bar(x + width/2, data["Vout_exp (V)"], width, label="Vout Eksperimental (V)", color="#DD8452")

plt.ylabel("Tensioni Vout (V)")
plt.title("Krahasimi i Tensionit Teorik dhe Eksperimental ne Dalje")
plt.xticks(x, ["Eksp 1", "Eksp 2", "Eksp 3a", "Eksp 3b"])
plt.legend()
plt.grid(axis="y", linestyle="--", alpha=0.7)
plt.tight_layout()
plt.show()
```

Nga grafiku dhe tabela e llogaritjeve verejme:

- **Ne Eksperimentin 1:** Kur $R_1 \approx R_2$ ($7.55\text{ k}\Omega$ dhe $7.78\text{ k}\Omega$), tensioni ne dalje $V_{out} = 7.50\text{ V}$ eshte shume afer gjysmes se tensionit hyres $V_{in} = 15.35\text{ V}$ ($V_{teori} = 7.79\text{ V}$).
- **Ne Eksperimentin 2:** Per rezistenca shume te medha ne rendin e Mega-Ohm ($3.27\text{ M}\Omega$ dhe $3.30\text{ M}\Omega$), vlera e matur $V_{out} = 6.71\text{ V}$ eshte me e ulet se vlera teorike ideale ($7.92\text{ V}$) per shkak te efektit ngarkues te rezistences se brendshme te multimetrit.
- **Ne Eksperimentin 3:** Shtimi i rezistences se ngarkesës $R_L$ ne paralel me $R_2$ ul rezistencen ekuivalente te deges se daljes.


# 6. Analiza e Gabimeve

Gabimi relativ llogaritet me formulen:

$$
\text{Gabimi} = \frac{|V_{exp} - V_{teori}|}{V_{teori}} \times 100\%
$$

Burimet kryesore te pasigurise dhe devijimeve ne matje jane:

- **Impedanca e brendshme e multimetrit (Efekti i ngarkeses):** Ne Eksperimentin 2, ku rezistencat jane ne rendin e $3.3\text{ M}\Omega$, rezistenca e brendshme e voltmetrit (zakonisht $10\text{ M}\Omega$) lidhet ne paralel me $R_2$, duke ulur ndjeshem tensionin e matur ($6.71\text{ V}$ kundrejt $7.92\text{ V}$).
- **Tolerancat e komponentëve dhe kontaktet:** Rezistenca e kontakteve ne vrimat e breadboard-it dhe luhatjet e vogla te tensionit te burimit DC.
- **Pasiguria e leximit:** Saktesia instrumentale e multimetrit digjital gjate matjes se tensionit dhe rezistences.


# 7. Diskutimi

1. **Sjellja e qarkut:** Ndaresi i tensionit funksionon ne menyre lineare dhe perputhet shume mire me parashikimet teorike kur rezistencat e qarkut jane ne rendin e kilo-Ohm ($	ext{k}\Omega$).
2. **Burimet e devijimit:** Devijimi me i dukshem shfaqet ne Eksperimentin 2 (me rezistenca $	ext{M}\Omega$), ku impedanca e voltmetrit nuk mund te konsiderohet me e pafundme ne krahasim me rezistencat e qarkut, si dhe ne Eksperimentin 3 ku lidhja e ngarkeses $R_L$ ndryshon ndarjen e tensionit.
3. **Permiresime:** Per te shmangur efektin ngarkues te instrumentit mates ne qarqe me rezistence te larte, rekomandohet perdorimi i nje buferi tensioni (voltage follower me amplifier operacional) para matjes me multimetër.
